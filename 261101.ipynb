모스부호 (1)

머쓱이는 친구에게 모스부호를 이용한 편지를 받았습니다. 그냥은 읽을 수 없어 이를 해독하는 프로그램을 만들려고 합니다. 문자열 letter가 매개변수로 주어질 때, letter를 영어 소문자로 바꾼 문자열을 return 하도록 solution 함수를 완성해보세요.
모스부호는 다음과 같습니다.

```python
# My Answer
morse = { 
    '.-':'a','-...':'b','-.-.':'c','-..':'d','.':'e','..-.':'f',
    '--.':'g','....':'h','..':'i','.---':'j','-.-':'k','.-..':'l',
    '--':'m','-.':'n','---':'o','.--.':'p','--.-':'q','.-.':'r',
    '...':'s','-':'t','..-':'u','...-':'v','.--':'w','-..-':'x',
    '-.--':'y','--..':'z'
}

def solution(letter):
    result = ""
    for i in letter.split(' '):
        result += morse[i]
    return result


print(solution(".... . .-.. .-.. ---"))
print(solution(".--. -.-- - .... --- -."))
```

```python
# Good Solution

def solution(letter):
    morse = {
        '.-':'a','-...':'b','-.-.':'c','-..':'d','.':'e','..-.':'f',
        '--.':'g','....':'h','..':'i','.---':'j','-.-':'k','.-..':'l',
        '--':'m','-.':'n','---':'o','.--.':'p','--.-':'q','.-.':'r',
        '...':'s','-':'t','..-':'u','...-':'v','.--':'w','-..-':'x',
        '-.--':'y','--..':'z'
    }

    return ''.join([morse[i] for i in letter.split(' ')])

print(solution(".... . .-.. .-.. ---"))
print(solution(".--. -.-- - .... --- -."))
```

Problem 2

팩토리얼

i 팩토리얼 (i!)은 1부터 i까지 정수의 곱을 의미합니다. 예를들어 5! = 5 * 4 * 3 * 2 * 1 = 120 입니다. 정수 n이 주어질 때 다음 조건을 만족하는 가장 큰 정수 i를 return 하도록 solution 함수를 완성해주세요.

```python
# My Answer

def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)

def solution(n):
    i = 0
    while n >= factorial(i):
        i += 1
    return i-1


print(solution(3628800))
print(solution(7))
```

```python
# Good Solution
def solution(n):
    i = 1
    factorial = 1
    
    # 팩토리얼 값이 n을 초과하기 전까지 반복
    while factorial <= n:
        i += 1
        factorial *= i
        
    # n을 초과한 시점의 i에서 1을 빼서 반환
    return i - 1
```

Problem 3

숨어있는 숫자의 덧셈 (2)

문자열 my_string이 매개변수로 주어집니다. my_string은 소문자, 대문자, 자연수로만 구성되어있습니다. my_string안의 자연수들의 합을 return하도록 solution 함수를 완성해주세요.

```python
# My Answer

def solution(my_string):
    answer = 0
    tmp = ""
    for i in my_string:
        if i.isdigit():
            tmp += i
        elif tmp != "":
            answer += int(tmp)
            tmp = ""

    if tmp != "":
        answer += int(tmp)
    return answer


print(solution("aAb1B2cC34oOp"))
print(solution("1a2b3c4d123Z"))
```

```python
# Good Solution

def solution(my_string):
    s = ''.join(i if i.isdigit() else ' ' for i in my_string)
    return sum(int(i) for i in s.split())
```
