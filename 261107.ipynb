Problem 2

외계어 사전

PROGRAMMERS-962 행성에 불시착한 우주비행사 머쓱이는 외계행성의 언어를 공부하려고 합니다. 알파벳이 담긴 배열 spell과 외계어 사전 dic이 매개변수로 주어집니다. spell에 담긴 알파벳을 한번씩만 모두 사용한 단어가 dic에 존재한다면 1, 존재하지 않는다면 2를 return하도록 solution 함수를 완성해주세요.

````python
# My Answer

def solution(spell, dic):
    for i in dic:
        cnt = 0
        for j in spell:
            if i.count(j) == 1:
                cnt += 1
        if cnt == len(spell):
            return 1

    return 2

print(solution(["p", "o", "s"],["sod", "eocd", "qixm", "adio", "soo"]))
print(solution(["z", "d", "x"],["def", "dww", "dzx", "loveaw"]))
print(solution(["s", "o", "m", "d"],["moos", "dzx", "smm", "sunmmo", "som"]))
````

작성하신 코드는 문제의 요구사항을 정확히 파악하여 잘 구현된 훌륭한 코드입니다. `count()` 메서드를 활용해 각 알파벳이 정확히 '한 번만' 쓰였는지 체크하는 논리가 아주 명확합니다.

다만, 한 가지 잠재적인 예외 케이스(Edge Case)를 보완하고, 파이썬의 특징을 살려 **더 간결하게 작성할 수 있는 방법**을 몇 가지 제안해 드립니다.

### 1. 작성하신 코드의 보완점 (길이 체크)

현재 코드는 만약 `spell`이 `["a", "b"]`이고 `dic` 안에 `"abc"`가 있다면, `"abc"` 안에 'a'가 1개, 'b'가 1개 있으므로 `cnt == 2`가 되어 `1`을 반환하게 됩니다.
즉, "spell 이외의 다른 알파벳이 섞여 있는 경우"에도 정답으로 처리될 위험이 있습니다.

단어의 길이가 `spell`의 길이와 완벽히 일치하는지 먼저 확인해 주면 더 견고한 코드가 됩니다.

```python
def solution(spell, dic):
    for i in dic:
        # 단어의 길이가 spell의 원소 개수와 같은지 먼저 확인 (다른 문자가 섞이는 것 방지)
        if len(i) == len(spell):
            cnt = 0
            for j in spell:
                if i.count(j) == 1:
                    cnt += 1
            if cnt == len(spell):
                return 1

    return 2

```

### 2. 더 파이썬다운(Pythonic) 코드 제안

이 문제는 "단어를 구성하는 알파벳의 종류와 개수가 완벽히 같은가(아나그램)"를 묻는 문제이므로, 정렬(`sorted`)이나 집합(`set`)을 사용하면 코드를 훨씬 짧고 효율적으로 작성할 수 있습니다.

#### 방법 A: `sorted()` 활용 (가장 직관적이고 권장하는 방법)

`spell`을 정렬한 결과와 `dic` 내의 단어를 정렬한 결과가 완벽히 똑같은지 비교합니다.

```python
def solution(spell, dic):
    # spell을 미리 정렬해 둡니다. (예: ["z", "d", "x"] -> ["d", "x", "z"])
    spell.sort()
    
    for word in dic:
        # 단어도 알파벳 순으로 정렬하여 비교합니다.
        if sorted(word) == spell:
            return 1
            
    return 2

```

#### 방법 B: 리스트 컴프리헨션(List Comprehension)을 활용한 한 줄 코딩

파이썬의 간결함을 극대화한 방식입니다.

```python
def solution(spell, dic):
    return 1 if any(sorted(word) == sorted(spell) for word in dic) else 2

```

모두 사용한 단어가 존재한다라는 거면 상관 없는 거 아니야?

질문하신 내용이 충분히 헷갈릴 수 있는 부분입니다. 한국어 어감상 "이 알파벳들을 다 쓴 단어면 다른 알파벳이 더 들어있어도 되는 거 아닌가?"라고 생각하시는 것이 아주 자연스럽습니다.

하지만 코딩 테스트에서 "주어진 알파벳을 한 번씩만 모두 사용해 단어를 만들어야 한다"는 조건은 보통 아나그램(Anagram)을 의미합니다.

쉽게 말해, **'spell'이라는 주머니에 들어있는 알파벳 블록들만 꺼내서 조립해 단어를 만들어야 한다**는 뜻입니다.
`spell`에 `["a", "b"]` 블록 2개만 주어졌다면, 내 마음대로 'c' 블록을 추가해서 `"abc"`를 만들 수는 없는 것과 같습니다. 즉, "다른 알파벳이 섞여 들어가면 안 된다"는 암묵적인 규칙이 포함되어 있습니다.

만약 문제의 의도가 "다른 알파벳이 섞여도 상관없다"였다면, 보통 "spell에 담긴 알파벳이 모두 포함된 단어"라고 다르게 표현합니다.

프로그래머스의 실제 채점 서버에 `spell = ["a", "b"]`인데 `dic = ["abc"]`인 함정 테스트 케이스가 없었다면 작성해주신 원본 코드로도 100점 통과가 되었을 수 있습니다. 하지만 출제 의도(주어진 알파벳만으로 구성된 단어 찾기)에 완벽하게 부합하고, 혹시 모를 깐깐한 테스트 케이스까지 방어하려면 길이가 같은지 확인하거나 정렬을 통해 비교하는 것이 훨씬 안전한 접근법입니다.

````python
# Good Solution
def solution(spell, dic):
    # spell을 미리 정렬해 둡니다. (예: ["z", "d", "x"] -> ["d", "x", "z"])
    spell.sort()
    
    for word in dic:
        # 단어도 알파벳 순으로 정렬하여 비교합니다.
        if sorted(word) == spell:
            return 1
            
    return 2
````

Problem 3

로그인 성공?

머쓱이는 프로그래머스에 로그인하려고 합니다. 머쓱이가 입력한 아이디와 패스워드가 담긴 배열 id_pw와 회원들의 정보가 담긴 2차원 배열 db가 주어질 때, 다음과 같이 로그인 성공, 실패에 따른 메시지를 return하도록 solution 함수를 완성해주세요.

아이디와 비밀번호가 모두 일치하는 회원정보가 있으면 "login"을 return합니다.
로그인이 실패했을 때 아이디가 일치하는 회원이 없다면 “fail”를, 아이디는 일치하지만 비밀번호가 일치하는 회원이 없다면 “wrong pw”를 return 합니다.

제한사항
회원들의 아이디는 문자열입니다.
회원들의 아이디는 알파벳 소문자와 숫자로만 이루어져 있습니다.
회원들의 패스워드는 숫자로 구성된 문자열입니다.
회원들의 비밀번호는 같을 수 있지만 아이디는 같을 수 없습니다.
id_pw의 길이는 2입니다.
id_pw와 db의 원소는 [아이디, 패스워드] 형태입니다.
1 ≤ 아이디의 길이 ≤ 15
1 ≤ 비밀번호의 길이 ≤ 6
1 ≤ db의 길이 ≤ 10
db의 원소의 길이는 2입니다.

````python
# My Answer

def solution(id_pw, db):

    for i, j in db:
        if i == id_pw[0]:
            return "login" if j == id_pw[1] else "wrong pw"
    
    return "fail"

print(solution(["meosseugi", "1234"],[["rardss", "123"], ["yyoom", "1234"], ["meosseugi", "1234"]]))
print(solution(["programmer01", "15789"],[["programmer02", "111111"], ["programmer00", "134"], ["programmer01", "1145"]]))
print(solution(["rabbit04", "98761"],[["jaja11", "98761"], ["krong0313", "29440"], ["rabbit00", "111333"]]))
````
